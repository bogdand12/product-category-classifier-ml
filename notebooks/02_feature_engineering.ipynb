# 02 - Feature Engineering

```
import pandas as pd
import re
df=pd.read_csv('../data/products.csv')
```

### title_length

```
df['title_length']=df['Product Title'].astype(str).apply(len)
```

### word_count

```
df['word_count']=df['Product Title'].astype(str).apply(lambda x: len(x.split()))
```

### has_numbers

```
df['has_numbers']=df['Product Title'].astype(str).apply(lambda x:int(bool(re.search(r'\d',x))))
```

### TF-IDF

```
from sklearn.feature_extraction.text import TfidfVectorizer
vec=TfidfVectorizer(stop_words='english',ngram_range=(1,2))
```
