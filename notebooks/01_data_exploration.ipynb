# 01 - Explorarea datelor pentru clasificarea automată a produselor

Acest notebook are rolul de a analiza setul de date inițial. Verificăm structura datelor, coloanele disponibile, valorile lipsă, distribuția categoriilor și câteva exemple de titluri de produse.

Scop: să înțelegem datele înainte de curățare și antrenarea modelelor.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

DATA_PATH = Path('products.csv')
if not DATA_PATH.exists():
    DATA_PATH = Path('/mnt/data/products.csv')

df = pd.read_csv(DATA_PATH)
df.columns = df.columns.str.strip()
df.head()
```

## 1. Dimensiunea setului de date și tipurile de coloane

```python
print(f"Număr de rânduri: {df.shape[0]}")
print(f"Număr de coloane: {df.shape[1]}")
df.info()
```

## 2. Verificarea valorilor lipsă

Valorile lipsă pot afecta antrenarea modelelor. Pentru clasificarea textului ne interesează în special coloanele `Product Title` și `Category Label`.

```python
missing = df.isna().sum().sort_values(ascending=False)
missing_percent = (df.isna().mean() * 100).round(2).sort_values(ascending=False)
pd.DataFrame({
    'Valori lipsă': missing,
    'Procent (%)': missing_percent
})
```

## 3. Distribuția produselor pe categorii

Această analiză arată dacă setul de date este echilibrat sau dacă unele categorii au mult mai multe produse decât altele.

```python
category_counts = df['Category Label'].astype(str).str.strip().value_counts()
category_counts
```

```python
plt.figure(figsize=(10, 6))
category_counts.plot(kind='bar')
plt.title('Distribuția produselor pe categorii')
plt.xlabel('Categorie')
plt.ylabel('Număr produse')
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()
```

## 4. Analiza lungimii titlurilor produselor

Titlul produsului este principala informație folosită pentru clasificare. Lungimea titlurilor poate influența calitatea reprezentării textului.

```python
df['title_length'] = df['Product Title'].astype(str).str.len()
df['word_count'] = df['Product Title'].astype(str).str.split().str.len()

df[['title_length', 'word_count']].describe()
```

```python
plt.figure(figsize=(8, 5))
df['word_count'].hist(bins=30)
plt.title('Distribuția numărului de cuvinte din titluri')
plt.xlabel('Număr de cuvinte')
plt.ylabel('Frecvență')
plt.tight_layout()
plt.show()
```

## 5. Exemple de produse pe categorii

```python
for category in category_counts.index[:10]:
    print('
Categorie:', category)
    examples = df.loc[df['Category Label'].astype(str).str.strip() == category, 'Product Title'].dropna().head(5)
    for item in examples:
        print(' -', item)
```

## Concluzii

- Setul de date conține titluri de produse și etichete de categorie.
- Pentru problema de clasificare, coloanele esențiale sunt `Product Title` și `Category Label`.
- Există valori lipsă, deci este necesară o etapă de curățare.
- Distribuția categoriilor trebuie analizată deoarece poate influența performanța modelelor.
